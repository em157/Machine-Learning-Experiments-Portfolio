## Executive Summary
- **Goal:** Adaptive thresholding for kNN image anomaly detection using a Multi-Armed Bandit (MAB).
- **Data:** 132 images → 2048-D embeddings.
- **Method:** kNN distance baseline + Thompson Sampling to pick thresholds.
- **Key Results:** Recall ↑ to 0.936; AUROC ↑ to 0.878; min_cost ↓ 13.6%.
- **Takeaway:** Bandit feedback calibrates decisions without retraining the model.
